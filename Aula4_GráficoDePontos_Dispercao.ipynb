```python
from matplotlib import pyplot as plt
```

```python
x=[1,3,5,7,9,11]
y=[5,9,7,3,8,2]

plt.scatter(x,y, color='navy', marker='*', s = 100)
plt.plot(x,y, linestyle='dashdot', color="pink", label="linha")

plt.title('Gráfico com Pontos')
plt.xlabel("Eixo x")
plt.ylabel("Eixo y")
plt.legend

plt.show()
```

```python
import matplotlib.pyplot as plt


x=[1,4,6]
rotulos = ['DNA', 'RNA', 'CITO']

plt.title('Gráfico de porcentagem')
plt.axis('equal')
plt.pie(x, labels = rotulos, autopct = '%1.1f%%')
plt.show()
```

figure = imagem a tabela.

*Para alterar o tamanho da figura:*

plt.figure(figsize= (largura, altura))

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x=[1,2,3,4]
y=[1,4,2,3]

t= np.arange(0.01,5.0,0.01)
s= np.sin(2 * np.pi *t)

plt.figure(figsize= (10,10))

plt.subplot(2,1,1)
plt.title("Primeiro Gráfico")
plt.plot(x,y, label = "Gráfico 1", color='navy')
plt.legend()

plt.subplot(2,1,2)
plt.title("Segundo Gráfico")
plt.plot(t,s, label = "Grafico 2", color='red')
plt.legend()

plt.show()
```
