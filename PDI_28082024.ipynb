```python
from google.colab.patches import cv2_imshow
import cv2
```

```python
#Carregar imagem, salvando em uma variável

img = cv2.imread('xray.jpg',0) # ,0 para escala em cinza
#img_cinza = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
# Verificar as dimensões da imagem original

altura, largura = img.shape[:2]
print(f"Dimensões Originais: {largura}x{altura}")
```

```python
# Definir novas dimensões

nova_largura = 500
nova_altura = int((nova_largura / largura) * altura)

img_red = cv2.resize(img, (nova_largura, nova_altura))
cv2_imshow(img_red)
```

```python
#Gerar bordas

img_borda = cv2.Laplacian(img_red, cv2.CV_8U)
cv2_imshow(img_borda)
```
