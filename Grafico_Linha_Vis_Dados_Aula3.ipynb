```python
df = pd.read_csv('drive/MyDrive/Colab Notebooks/Iris.csv', sep = ',')
df

plt.plot(df['SepalLengthCm'], df['PetalLengthCm'], color = 'green', linestyle = 'dotted');

plt.title("Iris Dataset");
plt.xlabel("Sepal Length Cm");
plt.ylabel("Petal Length Cm");

plt.show()
```

Para fazer gráficos de disperção, utilizar a função scatter.

```python
df = pd.read_csv('drive/MyDrive/Colab Notebooks/Iris.csv', sep = ',')
df

plt.scatter(df['SepalLengthCm'], df['PetalLengthCm'], color = 'green', linestyle = 'dotted')
plt.title("Iris Dataset");
plt.xlabel("Sepal Length Cm");
plt.ylabel("Petal Length Cm");

plt.show()
```

```python
df = pd.read_csv('drive/MyDrive/Colab Notebooks/Iris.csv', sep = ',')
df

plt.plot(df['SepalLengthCm'], df['PetalLengthCm'], color = 'green', linestyle = 'dotted', label = 'Gráfico 1');
plt.plot(df['SepalWidthCm'], df['PetalWidthCm'], color = 'pink', label = 'Gráfico 2');

plt.title("Dois gráficos | Iris Dataset");

plt.legend()
plt.show()
```
