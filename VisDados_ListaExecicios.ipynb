```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/iris1.csv', sep = ',')
df
```

```python
plt.figure(figsize=(22,8))

plt.subplot(1,3,1)

x = df['SepalLengthCm']
y = df['PetalLengthCm']

plt.plot(x,y, label = 'Sépala')
plt.plot(x,y, label = 'Pétala')

plt.title("Comparação de Sépala x Pétala")
plt.xlabel('Eixo x')
plt.ylabel('Eixo y')
plt.legend()

plt.subplot(1,3,2)
plt.bar(df['SepalLengthCm'], df['PetalLength'])

plt.subplot(1,3,3)
plt.scatter(df['SepalLengthCm'], df['Petal_Length'])

plt.show()
```

```python

x = df['SepalLengthCm']
y = df['PetalLengthCm']

plt.plot(x,y, label = 'Sépala')
plt.plot(x,y, label = 'Pétala')

plt.title("Comparação de Sépala x Pétala")
plt.xlabel('Eixo x')
plt.ylabel('Eixo y')
plt.legend()

plt.bar(df['SepalWidthCm'], df['PetalWidthCm'], color='navy')
plt.show()

plt.scatter(df['SepalWidthCm'], df['PetalWidthCm'], color='navy')
plt.show()
```

```python
plt.subplot(3,3,1)
plt.plot(df['SepalLengthCm'], df['PetalWidthCm'], color='black')

plt.subplot(3,3,2)
plt.barh(df['SepalLengthCm'], df['PetalWidthCm'], color='green')

plt.subplot(3,3,3)
plt.scatter(df['SepalLengthCm'], df['PetalWidthCm'], color='orange')


plt.show()

```
