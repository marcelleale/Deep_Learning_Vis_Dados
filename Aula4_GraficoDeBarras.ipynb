```python
from matplotlib import pyplot as plt

x=[1,3,5,7,9,11]
y=[5,9,7,3,8,2]

plt.plot(x,y, color='purple', label= "Linha")
plt.bar(x,y, color='pink', label= "Barra")

plt.title("Comparação de Gráficos de Barras e Linhas")
plt.xlabel('Eixo x')
plt.ylabel('Eixo y')
plt.legend()
```

```python
dados = [1,3,5,7,9,11]
nomes = ['RJ','SP','RS','MG','MA','CE']

plt.barh(nomes, dados, color='g', label = "Barra")

plt.title("Gráfico de Barra Horizontal")
plt.xlabel("Eixo x")
plt.ylabel("Eixo y")
plt.legend()

plt.show()
```
